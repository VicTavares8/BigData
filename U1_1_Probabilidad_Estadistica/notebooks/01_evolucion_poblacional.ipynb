# **Actividad 3 - Estadística Descriptiva**
## ***Propósito***
### Aplicar medidas estadísticas básicas a una variable poblacional. 

**Sobre la variable de cambio anual de población, calcular:**

```python
año = 1951
df_filtrado = df[df["Year"] == año]

variable = df_filtrado['Annual change in population']

print(f"--- Estadísticas descriptivas para el año {año} ---")
print(f"Media: {variable.mean()}")
print(f"Mediana: {variable.median()}")
print(f"Varianza: {variable.var()}")
print(f"Desviación estándar: {variable.std()}")
print(f"Mínimo: {variable.min()}")
print(f"Máximo: {variable.max()}")

entidad_min = df_filtrado[df_filtrado['Annual change in population'] == variable.min()]['Entity'].values[0]
entidad_max = df_filtrado[df_filtrado['Annual change in population'] == variable.max()]['Entity'].values[0]
print(f"Entidad con el valor mínimo: {entidad_min}")
print(f"Entidad con el valor máximo: {entidad_max}")
```

```python
#Filtrar por rango de años y por letra inicial del país
rango_años = (df['Year'] >= 2006) & (df['Year'] <= 2023)
letra_inicial_pais = df['Entity'].str.startswith('T')
df_filtrado = df.loc[rango_años & letra_inicial_pais]

variable = df_filtrado['Annual change in population']

print(f"Estadísticas del 2006 al 2023 para países que comienzan con la letra 'T':")
print(f"Media: {variable.mean()}")
print(f"Mediana: {variable.median()}")
print(f"Varianza: {variable.var()}")
print(f"Desviación estándar: {variable.std()}")
print(f"Mínimo: {variable.min()}")
print(f"Máximo: {variable.max()}")

entidad_min = df_filtrado[df_filtrado['Annual change in population'] == variable.min()]['Entity'].values[0]
entidad_max = df_filtrado[df_filtrado['Annual change in population'] == variable.max()]['Entity'].values[0]
print(f"Entidad con el valor mínimo: {entidad_min}")
print(f"Entidad con el valor máximo: {entidad_max}")
```
